```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully")
```

```python
orders_df = pd.read_csv("olist_orders_dataset.csv")

items_df = pd.read_csv("olist_order_items_dataset.csv")

products_df = pd.read_csv("olist_products_dataset.csv")

category_df = pd.read_csv("product_category_name_translation.csv")
```

```python
merged_df = items_df.merge(
    products_df,
    on="product_id",
    how="left"
)

merged_df = merged_df.merge(
    category_df,
    on="product_category_name",
    how="left"
)
```

```python
merged_df["revenue"] = (
    merged_df["price"] + merged_df["freight_value"]
)
```

```python
merged_df.head()
```

## Business Question

Which product categories generate the highest revenue?

```python
category_revenue = merged_df.groupby(
    "product_category_name_english"
)["revenue"].sum().sort_values(ascending=False)
```

```python
top_categories = category_revenue.head(10)

print(top_categories)
```

```python
plt.figure(figsize=(12,6))

top_categories.plot(kind="bar")

plt.title("Top 10 Revenue Generating Product Categories")

plt.xlabel("Product Category")

plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.show()
```

### Business Insights

1. Some product categories generate significantly higher revenue than others.
2. High-performing categories should receive more marketing and inventory focus.
3. Lower-performing categories may require promotional strategies or pricing improvements.
4. Revenue concentration indicates strong customer demand in specific categories.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

orders_df = pd.read_csv("olist_orders_dataset.csv")

items_df = pd.read_csv("olist_order_items_dataset.csv")

products_df = pd.read_csv("olist_products_dataset.csv")

category_df = pd.read_csv("product_category_name_translation.csv")

products_df = products_df.dropna()

merged_df = items_df.merge(
    products_df,
    on="product_id",
    how="left"
)

merged_df = merged_df.merge(
    category_df,
    on="product_category_name",
    how="left"
)

merged_df["revenue"] = (
    merged_df["price"] + merged_df["freight_value"]
)

print("Merged dataset ready")
```

```python
plt.figure(figsize=(8,5))

sns.histplot(merged_df["revenue"], bins=30, kde=True)

plt.title("Revenue Distribution")

plt.xlabel("Revenue")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    merged_df[["price", "freight_value", "revenue"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

## Final Conclusion

The analysis shows that certain product categories contribute significantly more revenue than others. Revenue patterns indicate strong customer demand in specific categories, which can help the business prioritize inventory, marketing campaigns, and seller partnerships.

Exploratory Data Analysis helped uncover important sales trends and revenue relationships that support data-driven business decisions.
