## Task 1 - Derivative and Slope Intuition

```python
# Function definition
def f(x):
    return x**2 + 2*x + 1

# Derivative function
def derivative(x):
    return 2*x + 2

# Chosen points
points = [-2, 0, 3]

# Print derivatives
for p in points:
    print(f"Derivative at x = {p}: {derivative(p)}")
```

```python
x = np.linspace(-5, 5, 200)
y = f(x)

plt.figure(figsize=(8,5))
plt.plot(x, y, label="f(x) = x² + 2x + 1")

# Tangent lines
for p in points:
    slope = derivative(p)
    tangent = slope*(x - p) + f(p)

    plt.plot(x, tangent, linestyle='--', label=f"Tangent at x={p}")

    plt.scatter(p, f(p))

plt.title("Function and Tangent Lines")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()

plt.show()
```

### Observations

1. The slope increases as x increases because the derivative becomes larger.
2. Positive slope means the function is increasing, while negative slope means the function is decreasing.
3. Near-zero slope indicates the function is close to a minimum or maximum point, which is important in ML optimization.

## Task 2 - Gradient Descent

```python
# Cost function
def cost_function(x):
    return x**2

# Derivative of cost function
def gradient(x):
    return 2*x
```

```python
# Initial value
x = 8

# Learning rate
learning_rate = 0.1

# Store values
x_history = []
cost_history = []

# Gradient Descent
for i in range(20):

    x_history.append(x)

    cost = cost_function(x)
    cost_history.append(cost)

    x = x - learning_rate * gradient(x)

    print(f"Iteration {i+1}: x = {x:.4f}, cost = {cost:.4f}")
```

```python
plt.figure(figsize=(8,5))

plt.plot(cost_history, marker='o')

plt.title("Gradient Descent Optimization")
plt.xlabel("Iterations")
plt.ylabel("Cost")

plt.show()
```

### Observations

1. The cost decreases after every iteration.
2. Gradient descent moves towards the minimum value of the function.
3. Learning rate controls the speed of optimization.
4. Very large learning rates may overshoot the minimum.

## Task 3 - Probability Distributions

```python
# Normal distribution
normal_data = np.random.normal(loc=50, scale=10, size=1000)

# Uniform distribution
uniform_data = np.random.uniform(low=0, high=100, size=1000)
```

```python
plt.figure(figsize=(8,5))

sns.histplot(normal_data, bins=30, kde=True)

plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(uniform_data, bins=30, kde=True)

plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

### Observations

1. Normal distribution forms a bell-shaped curve.
2. Most values in normal distribution are concentrated near the mean.
3. Uniform distribution spreads values evenly across the range.
4. Probability distributions are important in statistics and machine learning.
