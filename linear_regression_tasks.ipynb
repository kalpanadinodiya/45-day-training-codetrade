```python
X_single = housing_df[["median_income"]]

y = housing_df["median_house_value"]
```

```python
split_results = []
```

```python
splits = [0.2, 0.3, 0.4]
```

```python
for split in splits:

    X_train, X_test, y_train, y_test = train_test_split(
        X_multi,
        y,
        test_size=split,
        random_state=42
    )

    model = LinearRegression()

    model.fit(X_train, y_train)

    y_train_pred = model.predict(X_train)

    y_test_pred = model.predict(X_test)

    train_mse = mean_squared_error(y_train, y_train_pred)

    test_mse = mean_squared_error(y_test, y_test_pred)

    train_rmse = np.sqrt(train_mse)

    test_rmse = np.sqrt(test_mse)

    train_mae = mean_absolute_error(y_train, y_train_pred)

    test_mae = mean_absolute_error(y_test, y_test_pred)

    train_r2 = r2_score(y_train, y_train_pred)

    test_r2 = r2_score(y_test, y_test_pred)

    split_results.append([
        split,
        train_mse,
        test_mse,
        train_rmse,
        test_rmse,
        train_mae,
        test_mae,
        train_r2,
        test_r2
    ])
```

```python
split_df = pd.DataFrame(
    split_results,
    columns=[
        "Test Size",
        "Train MSE",
        "Test MSE",
        "Train RMSE",
        "Test RMSE",
        "Train MAE",
        "Test MAE",
        "Train R2",
        "Test R2"
    ]
)

split_df
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    split_df["Test Size"],
    split_df["Test R2"],
    marker="o"
)

plt.title("Test R2 Score Across Different Splits")

plt.xlabel("Test Size")

plt.ylabel("Test R2 Score")

plt.show()
```

## Observation

The model performance remained relatively stable across different train-test splits. Comparing train and test metrics helps evaluate whether the model generalizes well to unseen data.

```python
manual_mse = np.mean((y_test_B - y_pred_B) ** 2)

manual_rmse = np.sqrt(manual_mse)

manual_mae = np.mean(np.abs(y_test_B - y_pred_B))

manual_r2 = 1 - (
    np.sum((y_test_B - y_pred_B) ** 2)
    /
    np.sum((y_test_B - np.mean(y_test_B)) ** 2)
)

print("Manual MSE:", manual_mse)

print("Manual RMSE:", manual_rmse)

print("Manual MAE:", manual_mae)

print("Manual R2:", manual_r2)
```

```python
from sklearn.metrics import median_absolute_error
```

```python
median_ae = median_absolute_error(
    y_test_B,
    y_pred_B
)

print("Median Absolute Error:", median_ae)
```

## Final Observation

After adding artificial large prediction errors, MSE and RMSE increased significantly compared to MAE. This shows that MSE and RMSE react more strongly to large errors because they square the error values.
